```python
# ============================================================
# BUILD LLM FROM SCRATCH - FINAL OPTIMIZED VERSION
# OPTIMIZED FOR GOOGLE COLAB T4 GPU
# ============================================================

# ============================================================
# 1. INSTALL LIBRARIES
# ============================================================

!pip install -q tiktoken torchinfo tqdm

# ============================================================
# 2. IMPORT LIBRARIES
# ============================================================

import os
import gc
import glob
import math
import torch
import tiktoken
import numpy as np
import pandas as pd
import torch.nn as nn
import matplotlib.pyplot as plt

from tqdm import tqdm
from torch.utils.data import Dataset, DataLoader
from torchinfo import summary

# ============================================================
# 3. GPU CHECK
# ============================================================

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

print("Using Device:", DEVICE)

if DEVICE == "cuda":
    print("GPU:", torch.cuda.get_device_name(0))

# ============================================================
# 4. LOAD ALL TXT FILES
# ============================================================

text = ""

txt_files = glob.glob("/content/*.txt")

print("\nFILES FOUND:")
print(txt_files)

for file_name in txt_files:

    with open(file_name, "r", encoding="utf-8", errors="ignore") as f:

        text += f.read() + "\n"

print("\nTotal Characters:", len(text))

print("\nSample Text:\n")
print(text[:1000])

# ============================================================
# 5. TOKENIZATION
# ============================================================

encoder = tiktoken.get_encoding("gpt2")

tokens = encoder.encode(text)

print("\nTotal Tokens:", len(tokens))

# ============================================================
# 6. TRAIN VALIDATION SPLIT
# ============================================================

split_idx = int(0.9 * len(tokens))

train_tokens = tokens[:split_idx]
val_tokens = tokens[split_idx:]

print("Train Tokens:", len(train_tokens))
print("Validation Tokens:", len(val_tokens))

# ============================================================
# 7. HYPERPARAMETERS
# OPTIMIZED FOR T4 GPU
# ============================================================

BATCH_SIZE = 8
BLOCK_SIZE = 64

VOCAB_SIZE = 50257

EMBED_DIM = 128

NUM_HEADS = 2
NUM_LAYERS = 2

DROPOUT = 0.1

LEARNING_RATE = 1e-3
EPOCHS = 5

# ============================================================
# 8. DATASET CLASS
# ============================================================

class GPTDataset(Dataset):

    def __init__(self, tokens, block_size):

        self.tokens = tokens
        self.block_size = block_size

    def __len__(self):

        return len(self.tokens) - self.block_size

    def __getitem__(self, idx):

        x = torch.tensor(
            self.tokens[idx:idx+self.block_size],
            dtype=torch.long
        )

        y = torch.tensor(
            self.tokens[idx+1:idx+self.block_size+1],
            dtype=torch.long
        )

        return x, y

# ============================================================
# 9. DATALOADERS
# ============================================================

train_dataset = GPTDataset(train_tokens, BLOCK_SIZE)

val_dataset = GPTDataset(val_tokens, BLOCK_SIZE)

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    pin_memory=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    pin_memory=True
)

# ============================================================
# 10. MULTI HEAD ATTENTION
# ============================================================

class MultiHeadAttention(nn.Module):

    def __init__(self, embed_dim, num_heads, dropout=0.1):

        super().__init__()

        assert embed_dim % num_heads == 0

        self.embed_dim = embed_dim
        self.num_heads = num_heads

        self.head_dim = embed_dim // num_heads

        self.query = nn.Linear(embed_dim, embed_dim)
        self.key = nn.Linear(embed_dim, embed_dim)
        self.value = nn.Linear(embed_dim, embed_dim)

        self.fc_out = nn.Linear(embed_dim, embed_dim)

        self.dropout = nn.Dropout(dropout)

    def forward(self, x):

        B, T, C = x.shape

        Q = self.query(x)
        K = self.key(x)
        V = self.value(x)

        Q = Q.view(B, T, self.num_heads, self.head_dim).transpose(1,2)
        K = K.view(B, T, self.num_heads, self.head_dim).transpose(1,2)
        V = V.view(B, T, self.num_heads, self.head_dim).transpose(1,2)

        attention = (Q @ K.transpose(-2,-1)) / math.sqrt(self.head_dim)

        mask = torch.tril(torch.ones(T,T, device=x.device))

        attention = attention.masked_fill(mask == 0, float('-inf'))

        attention = torch.softmax(attention, dim=-1)

        attention = self.dropout(attention)

        out = attention @ V

        out = out.transpose(1,2).contiguous().view(B,T,C)

        out = self.fc_out(out)

        return out

# ============================================================
# 11. FEED FORWARD NETWORK
# ============================================================

class FeedForward(nn.Module):

    def __init__(self, embed_dim, dropout=0.1):

        super().__init__()

        self.net = nn.Sequential(

            nn.Linear(embed_dim, 4 * embed_dim),

            nn.ReLU(),

            nn.Linear(4 * embed_dim, embed_dim),

            nn.Dropout(dropout)
        )

    def forward(self, x):

        return self.net(x)

# ============================================================
# 12. TRANSFORMER BLOCK
# ============================================================

class TransformerBlock(nn.Module):

    def __init__(
        self,
        embed_dim,
        num_heads,
        dropout=0.1,
        use_norm=True,
        use_residual=True,
        use_ffn=True
    ):

        super().__init__()

        self.use_norm = use_norm
        self.use_residual = use_residual
        self.use_ffn = use_ffn

        self.attention = MultiHeadAttention(
            embed_dim,
            num_heads,
            dropout
        )

        self.ffn = FeedForward(embed_dim, dropout)

        self.norm1 = nn.LayerNorm(embed_dim)
        self.norm2 = nn.LayerNorm(embed_dim)

    def forward(self, x):

        attn = self.attention(x)

        if self.use_residual:
            x = x + attn
        else:
            x = attn

        if self.use_norm:
            x = self.norm1(x)

        if self.use_ffn:

            ffn_out = self.ffn(x)

            if self.use_residual:
                x = x + ffn_out
            else:
                x = ffn_out

            if self.use_norm:
                x = self.norm2(x)

        return x

# ============================================================
# 13. GPT MODEL
# ============================================================

class GPTModel(nn.Module):

    def __init__(
        self,
        vocab_size,
        embed_dim,
        block_size,
        num_heads,
        num_layers,
        dropout=0.1,
        use_norm=True,
        use_residual=True,
        use_ffn=True
    ):

        super().__init__()

        self.token_embedding = nn.Embedding(
            vocab_size,
            embed_dim
        )

        self.position_embedding = nn.Embedding(
            block_size,
            embed_dim
        )

        self.blocks = nn.Sequential(*[

            TransformerBlock(
                embed_dim,
                num_heads,
                dropout,
                use_norm,
                use_residual,
                use_ffn
            )

            for _ in range(num_layers)

        ])

        self.ln_f = nn.LayerNorm(embed_dim)

        self.head = nn.Linear(embed_dim, vocab_size)

        self.dropout = nn.Dropout(dropout)

    def forward(self, x, targets=None):

        B, T = x.shape

        token_emb = self.token_embedding(x)

        positions = torch.arange(0, T, device=x.device)

        pos_emb = self.position_embedding(positions)

        out = token_emb + pos_emb

        out = self.dropout(out)

        out = self.blocks(out)

        out = self.ln_f(out)

        logits = self.head(out)

        loss = None

        if targets is not None:

            B, T, C = logits.shape

            logits = logits.view(B*T, C)

            targets = targets.view(B*T)

            loss = nn.functional.cross_entropy(
                logits,
                targets
            )

        return logits, loss

# ============================================================
# 14. TRAIN FUNCTION
# ============================================================

def train_model(
    epochs=5,
    learning_rate=1e-3,
    num_layers=2,
    num_heads=2,
    use_norm=True,
    use_residual=True,
    use_ffn=True
):

    torch.cuda.empty_cache()
    gc.collect()

    model = GPTModel(
        vocab_size=VOCAB_SIZE,
        embed_dim=EMBED_DIM,
        block_size=BLOCK_SIZE,
        num_heads=num_heads,
        num_layers=num_layers,
        dropout=DROPOUT,
        use_norm=use_norm,
        use_residual=use_residual,
        use_ffn=use_ffn
    ).to(DEVICE)

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=learning_rate
    )

    scaler = torch.cuda.amp.GradScaler()

    train_losses = []
    val_losses = []

    for epoch in range(epochs):

        model.train()

        total_train_loss = 0

        loop = tqdm(train_loader)

        for x, y in loop:

            x = x.to(DEVICE)
            y = y.to(DEVICE)

            optimizer.zero_grad()

            with torch.cuda.amp.autocast():

                _, loss = model(x, y)

            scaler.scale(loss).backward()

            scaler.step(optimizer)

            scaler.update()

            total_train_loss += loss.item()

            loop.set_description(f"Epoch {epoch+1}")

            loop.set_postfix(loss=loss.item())

        avg_train_loss = total_train_loss / len(train_loader)

        model.eval()

        total_val_loss = 0

        with torch.no_grad():

            for x, y in val_loader:

                x = x.to(DEVICE)
                y = y.to(DEVICE)

                with torch.cuda.amp.autocast():

                    _, loss = model(x, y)

                total_val_loss += loss.item()

        avg_val_loss = total_val_loss / len(val_loader)

        train_losses.append(avg_train_loss)
        val_losses.append(avg_val_loss)

        print(
            f"\nEpoch {epoch+1}/{epochs}"
            f"\nTrain Loss: {avg_train_loss:.4f}"
            f"\nValidation Loss: {avg_val_loss:.4f}\n"
        )

    return model, train_losses, val_losses

# ============================================================
# 15. PLOT FUNCTION
# ============================================================

def plot_losses(train_loss, val_loss, title):

    plt.figure(figsize=(8,5))

    plt.plot(train_loss, marker='o', label='Train Loss')

    plt.plot(val_loss, marker='o', label='Validation Loss')

    plt.xlabel("Epoch")

    plt.ylabel("Loss")

    plt.title(title)

    plt.legend()

    plt.grid(True)

    plt.show()

# ============================================================
# 16. BASELINE TRAINING
# ============================================================

baseline_model, baseline_train_loss, baseline_val_loss = train_model(
    epochs=5,
    learning_rate=1e-3,
    num_layers=2,
    num_heads=2
)

plot_losses(
    baseline_train_loss,
    baseline_val_loss,
    "Baseline Model"
)

# ============================================================
# 17. EXPERIMENT 1
# LEARNING RATE + EPOCHS
# ============================================================

learning_rates = [1e-4, 1e-3]

epoch_values = [3, 5]

experiment1_results = {}

for lr in learning_rates:

    for ep in epoch_values:

        print(f"\nRunning LR={lr}, Epochs={ep}")

        model, train_loss, val_loss = train_model(
            epochs=ep,
            learning_rate=lr,
            num_layers=2,
            num_heads=2
        )

        experiment1_results[(lr, ep)] = {
            "train": train_loss,
            "val": val_loss
        }

        plot_losses(
            train_loss,
            val_loss,
            f"LR={lr}, Epochs={ep}"
        )

# ============================================================
# 18. EXPERIMENT 2
# TRANSFORMER LAYERS
# ============================================================

layer_configs = [1, 3, 5]

layer_results = {}

for layers in layer_configs:

    print(f"\nRunning {layers} Layers")

    model, train_loss, val_loss = train_model(
        epochs=5,
        learning_rate=1e-3,
        num_layers=layers,
        num_heads=2
    )

    layer_results[layers] = {
        "train": train_loss,
        "val": val_loss
    }

    plot_losses(
        train_loss,
        val_loss,
        f"{layers} Transformer Layers"
    )

# ============================================================
# 19. EXPERIMENT 3
# ATTENTION HEADS
# ============================================================

head_configs = [1, 2, 4]

head_results = {}

for heads in head_configs:

    print(f"\nRunning {heads} Attention Heads")

    model, train_loss, val_loss = train_model(
        epochs=5,
        learning_rate=1e-3,
        num_layers=2,
        num_heads=heads
    )

    head_results[heads] = {
        "train": train_loss,
        "val": val_loss
    }

    plot_losses(
        train_loss,
        val_loss,
        f"{heads} Attention Heads"
    )

# ============================================================
# 20. EXPERIMENT 4
# ABLATION STUDIES
# ============================================================

# ------------------------------------------------------------
# A. REMOVE NORMALIZATION
# ------------------------------------------------------------

norm_model, norm_train, norm_val = train_model(
    epochs=5,
    learning_rate=1e-3,
    num_layers=2,
    num_heads=2,
    use_norm=False
)

plot_losses(
    norm_train,
    norm_val,
    "Without Layer Normalization"
)

# ------------------------------------------------------------
# B. REMOVE RESIDUAL CONNECTIONS
# ------------------------------------------------------------

res_model, res_train, res_val = train_model(
    epochs=5,
    learning_rate=1e-3,
    num_layers=2,
    num_heads=2,
    use_residual=False
)

plot_losses(
    res_train,
    res_val,
    "Without Residual Connections"
)

# ------------------------------------------------------------
# C. REMOVE FEED FORWARD NETWORK
# ------------------------------------------------------------

ffn_model, ffn_train, ffn_val = train_model(
    epochs=5,
    learning_rate=1e-3,
    num_layers=2,
    num_heads=2,
    use_ffn=False
)

plot_losses(
    ffn_train,
    ffn_val,
    "Without Feed Forward Network"
)

# ============================================================
# 21. TEXT GENERATION
# ============================================================

def generate_text(model, start_text, max_tokens=100):

    model.eval()

    tokens = encoder.encode(start_text)

    x = torch.tensor(tokens).unsqueeze(0).to(DEVICE)

    for _ in range(max_tokens):

        x_cond = x[:, -BLOCK_SIZE:]

        with torch.no_grad():

            logits, _ = model(x_cond)

        logits = logits[:, -1, :]

        probs = torch.softmax(logits, dim=-1)

        next_token = torch.multinomial(probs, num_samples=1)

        x = torch.cat((x, next_token), dim=1)

    generated = encoder.decode(x[0].tolist())

    return generated

# ============================================================
# 22. GENERATE SAMPLE TEXT
# ============================================================

sample_text = generate_text(
    baseline_model,
    "Once upon a time",
    max_tokens=100
)

print("\nGENERATED TEXT:\n")
print(sample_text)

# ============================================================
# 23. SAVE MODEL
# ============================================================

torch.save(
    baseline_model.state_dict(),
    "baseline_gpt_model.pth"
)

print("\nModel Saved Successfully")

# ============================================================
# 24. SUMMARY TABLE
# ============================================================

summary_data = {

    "Experiment": [

        "Baseline",
        "Low Learning Rate",
        "Higher Layers",
        "More Attention Heads",
        "No LayerNorm",
        "No Residual",
        "No FFN"
    ],

    "Observation": [

        "Balanced Learning",
        "Stable but Slow",
        "Better Representation",
        "Richer Context Learning",
        "Unstable Training",
        "Gradient Flow Problems",
        "Reduced Learning Power"
    ]
}

summary_df = pd.DataFrame(summary_data)

print("\nSUMMARY TABLE:\n")

print(summary_df)

# ============================================================
# END OF FINAL OPTIMIZED CODE
# ============================================================
```

```python
# ============================================================
# IMPORTS
# ============================================================

import os
import gc
import glob
import math
import torch
import tiktoken
import numpy as np
import pandas as pd
import torch.nn as nn
import matplotlib.pyplot as plt

from tqdm import tqdm
from torch.utils.data import Dataset, DataLoader

# ============================================================
# GPU CHECK
# ============================================================

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

print("Using Device:", DEVICE)

if DEVICE == "cuda":
    print("GPU:", torch.cuda.get_device_name(0))

# ============================================================
# LOAD TXT FILES
# ============================================================

text = ""

txt_files = glob.glob("/content/*.txt")

print("\nFILES FOUND:")
print(txt_files)

for file_name in txt_files:

    with open(file_name, "r", encoding="utf-8", errors="ignore") as f:

        text += f.read() + "\n"

print("\nTotal Characters:", len(text))

# ============================================================
# TOKENIZATION
# ============================================================

encoder = tiktoken.get_encoding("gpt2")

tokens = encoder.encode(text)

print("\nTotal Tokens:", len(tokens))

# ============================================================
# TRAIN VALIDATION SPLIT
# ============================================================

split_idx = int(0.9 * len(tokens))

train_tokens = tokens[:split_idx]
val_tokens = tokens[split_idx:]

print("Train Tokens:", len(train_tokens))
print("Validation Tokens:", len(val_tokens))

# ============================================================
# HYPERPARAMETERS
# ============================================================

BATCH_SIZE = 8
BLOCK_SIZE = 64

VOCAB_SIZE = 50257

EMBED_DIM = 128

NUM_HEADS = 2
NUM_LAYERS = 2

DROPOUT = 0.1

# ============================================================
# DATASET CLASS
# ============================================================

class GPTDataset(Dataset):

    def __init__(self, tokens, block_size):

        self.tokens = tokens
        self.block_size = block_size

    def __len__(self):

        return len(self.tokens) - self.block_size

    def __getitem__(self, idx):

        x = torch.tensor(
            self.tokens[idx:idx+self.block_size],
            dtype=torch.long
        )

        y = torch.tensor(
            self.tokens[idx+1:idx+self.block_size+1],
            dtype=torch.long
        )

        return x, y

# ============================================================
# DATALOADERS
# ============================================================

train_dataset = GPTDataset(train_tokens, BLOCK_SIZE)

val_dataset = GPTDataset(val_tokens, BLOCK_SIZE)

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    pin_memory=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    pin_memory=True
)

# ============================================================
# MULTI HEAD ATTENTION
# ============================================================

class MultiHeadAttention(nn.Module):

    def __init__(self, embed_dim, num_heads, dropout=0.1):

        super().__init__()

        assert embed_dim % num_heads == 0

        self.embed_dim = embed_dim
        self.num_heads = num_heads

        self.head_dim = embed_dim // num_heads

        self.query = nn.Linear(embed_dim, embed_dim)
        self.key = nn.Linear(embed_dim, embed_dim)
        self.value = nn.Linear(embed_dim, embed_dim)

        self.fc_out = nn.Linear(embed_dim, embed_dim)

        self.dropout = nn.Dropout(dropout)

    def forward(self, x):

        B, T, C = x.shape

        Q = self.query(x)
        K = self.key(x)
        V = self.value(x)

        Q = Q.view(B, T, self.num_heads, self.head_dim).transpose(1,2)
        K = K.view(B, T, self.num_heads, self.head_dim).transpose(1,2)
        V = V.view(B, T, self.num_heads, self.head_dim).transpose(1,2)

        attention = (Q @ K.transpose(-2,-1)) / math.sqrt(self.head_dim)

        mask = torch.tril(torch.ones(T,T, device=x.device))

        attention = attention.masked_fill(mask == 0, float('-inf'))

        attention = torch.softmax(attention, dim=-1)

        attention = self.dropout(attention)

        out = attention @ V

        out = out.transpose(1,2).contiguous().view(B,T,C)

        out = self.fc_out(out)

        return out

# ============================================================
# FEED FORWARD NETWORK
# ============================================================

class FeedForward(nn.Module):

    def __init__(self, embed_dim, dropout=0.1):

        super().__init__()

        self.net = nn.Sequential(

            nn.Linear(embed_dim, 4 * embed_dim),

            nn.ReLU(),

            nn.Linear(4 * embed_dim, embed_dim),

            nn.Dropout(dropout)
        )

    def forward(self, x):

        return self.net(x)

# ============================================================
# TRANSFORMER BLOCK
# ============================================================

class TransformerBlock(nn.Module):

    def __init__(
        self,
        embed_dim,
        num_heads,
        dropout=0.1,
        use_norm=True,
        use_residual=True,
        use_ffn=True
    ):

        super().__init__()

        self.use_norm = use_norm
        self.use_residual = use_residual
        self.use_ffn = use_ffn

        self.attention = MultiHeadAttention(
            embed_dim,
            num_heads,
            dropout
        )

        self.ffn = FeedForward(embed_dim, dropout)

        self.norm1 = nn.LayerNorm(embed_dim)
        self.norm2 = nn.LayerNorm(embed_dim)

    def forward(self, x):

        attn = self.attention(x)

        if self.use_residual:
            x = x + attn
        else:
            x = attn

        if self.use_norm:
            x = self.norm1(x)

        if self.use_ffn:

            ffn_out = self.ffn(x)

            if self.use_residual:
                x = x + ffn_out
            else:
                x = ffn_out

            if self.use_norm:
                x = self.norm2(x)

        return x

# ============================================================
# GPT MODEL
# ============================================================

class GPTModel(nn.Module):

    def __init__(
        self,
        vocab_size,
        embed_dim,
        block_size,
        num_heads,
        num_layers,
        dropout=0.1,
        use_norm=True,
        use_residual=True,
        use_ffn=True
    ):

        super().__init__()

        self.token_embedding = nn.Embedding(
            vocab_size,
            embed_dim
        )

        self.position_embedding = nn.Embedding(
            block_size,
            embed_dim
        )

        self.blocks = nn.Sequential(*[

            TransformerBlock(
                embed_dim,
                num_heads,
                dropout,
                use_norm,
                use_residual,
                use_ffn
            )

            for _ in range(num_layers)

        ])

        self.ln_f = nn.LayerNorm(embed_dim)

        self.head = nn.Linear(embed_dim, vocab_size)

        self.dropout = nn.Dropout(dropout)

    def forward(self, x, targets=None):

        B, T = x.shape

        token_emb = self.token_embedding(x)

        positions = torch.arange(0, T, device=x.device)

        pos_emb = self.position_embedding(positions)

        out = token_emb + pos_emb

        out = self.dropout(out)

        out = self.blocks(out)

        out = self.ln_f(out)

        logits = self.head(out)

        loss = None

        if targets is not None:

            B, T, C = logits.shape

            logits = logits.view(B*T, C)

            targets = targets.view(B*T)

            loss = nn.functional.cross_entropy(
                logits,
                targets
            )

        return logits, loss

# ============================================================
# TRAIN FUNCTION
# ============================================================

def train_model(
    epochs=3,
    learning_rate=1e-3,
    num_layers=2,
    num_heads=2,
    use_norm=True,
    use_residual=True,
    use_ffn=True
):

    torch.cuda.empty_cache()
    gc.collect()

    model = GPTModel(
        vocab_size=VOCAB_SIZE,
        embed_dim=EMBED_DIM,
        block_size=BLOCK_SIZE,
        num_heads=num_heads,
        num_layers=num_layers,
        dropout=DROPOUT,
        use_norm=use_norm,
        use_residual=use_residual,
        use_ffn=use_ffn
    ).to(DEVICE)

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=learning_rate
    )

    scaler = torch.cuda.amp.GradScaler()

    train_losses = []
    val_losses = []

    for epoch in range(epochs):

        model.train()

        total_train_loss = 0

        loop = tqdm(train_loader)

        for x, y in loop:

            x = x.to(DEVICE)
            y = y.to(DEVICE)

            optimizer.zero_grad()

            with torch.cuda.amp.autocast():

                _, loss = model(x, y)

            scaler.scale(loss).backward()

            scaler.step(optimizer)

            scaler.update()

            total_train_loss += loss.item()

            loop.set_description(f"Epoch {epoch+1}")

            loop.set_postfix(loss=loss.item())

        avg_train_loss = total_train_loss / len(train_loader)

        model.eval()

        total_val_loss = 0

        with torch.no_grad():

            for x, y in val_loader:

                x = x.to(DEVICE)
                y = y.to(DEVICE)

                with torch.cuda.amp.autocast():

                    _, loss = model(x, y)

                total_val_loss += loss.item()

        avg_val_loss = total_val_loss / len(val_loader)

        train_losses.append(avg_train_loss)
        val_losses.append(avg_val_loss)

        print(
            f"\nEpoch {epoch+1}/{epochs}"
            f"\nTrain Loss: {avg_train_loss:.4f}"
            f"\nValidation Loss: {avg_val_loss:.4f}\n"
        )

    return model, train_losses, val_losses
```

```python
# ============================================================
# EXPERIMENT 2
# VARYING TRANSFORMER LAYERS
# ============================================================

# Using fewer epochs for faster execution on T4 GPU

layer_configs = [1, 3, 5]

layer_results = {}

for layers in layer_configs:

    print(f"\n================================================")
    print(f"TRAINING MODEL WITH {layers} TRANSFORMER LAYERS")
    print(f"================================================\n")

    torch.cuda.empty_cache()
    gc.collect()

    model, train_loss, val_loss = train_model(
        epochs=3,
        learning_rate=1e-3,
        num_layers=layers,
        num_heads=2
    )

    layer_results[layers] = {

        "train_loss": train_loss,
        "val_loss": val_loss
    }

    # --------------------------------------------------------
    # PLOT LOSSES
    # --------------------------------------------------------

    plt.figure(figsize=(8,5))

    plt.plot(
        train_loss,
        marker='o',
        label='Train Loss'
    )

    plt.plot(
        val_loss,
        marker='o',
        label='Validation Loss'
    )

    plt.xlabel("Epoch")

    plt.ylabel("Loss")

    plt.title(f"{layers} Transformer Layers")

    plt.legend()

    plt.grid(True)

    plt.show()

# ============================================================
# FINAL COMPARISON PLOT
# ============================================================

plt.figure(figsize=(10,6))

for layers in layer_configs:

    plt.plot(
        layer_results[layers]["val_loss"],
        marker='o',
        label=f'{layers} Layers'
    )

plt.xlabel("Epoch")

plt.ylabel("Validation Loss")

plt.title("Comparison of Transformer Layers")

plt.legend()

plt.grid(True)

plt.show()

# ============================================================
# PRINT FINAL RESULTS
# ============================================================

print("\nFINAL RESULTS\n")

for layers in layer_configs:

    final_train = layer_results[layers]["train_loss"][-1]

    final_val = layer_results[layers]["val_loss"][-1]

    print(f"{layers} Layers")
    print(f"Final Train Loss: {final_train:.4f}")
    print(f"Final Validation Loss: {final_val:.4f}")
    print("-" * 40)

# ============================================================
# EXPERIMENT 2 ANALYSIS
# ============================================================

print("\nINFERENCE:\n")

print("1 Layer  -> Fast training but limited learning capacity")

print("3 Layers -> Better contextual understanding and balance")

print("5 Layers -> Higher representational power but increased overfitting risk")
```
