```python
# ============================================================
# IMPORTS
# ============================================================

import os
import gc
import glob
import math
import torch
import tiktoken
import numpy as np
import pandas as pd
import torch.nn as nn
import matplotlib.pyplot as plt

from tqdm import tqdm
from torch.utils.data import Dataset, DataLoader

# ============================================================
# GPU CHECK
# ============================================================

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

print("Using Device:", DEVICE)

if DEVICE == "cuda":
    print("GPU:", torch.cuda.get_device_name(0))

# ============================================================
# LOAD TXT FILES
# ============================================================

text = ""

txt_files = glob.glob("/content/*.txt")

print("\nFILES FOUND:")
print(txt_files)

for file_name in txt_files:

    with open(file_name, "r", encoding="utf-8", errors="ignore") as f:

        text += f.read() + "\n"

print("\nTotal Characters:", len(text))

# ============================================================
# TOKENIZATION
# ============================================================

encoder = tiktoken.get_encoding("gpt2")

tokens = encoder.encode(text)

print("\nTotal Tokens:", len(tokens))

# ============================================================
# TRAIN VALIDATION SPLIT
# ============================================================

split_idx = int(0.9 * len(tokens))

train_tokens = tokens[:split_idx]
val_tokens = tokens[split_idx:]

print("Train Tokens:", len(train_tokens))
print("Validation Tokens:", len(val_tokens))

# ============================================================
# HYPERPARAMETERS
# ============================================================

BATCH_SIZE = 8
BLOCK_SIZE = 64

VOCAB_SIZE = 50257

EMBED_DIM = 128

NUM_HEADS = 2
NUM_LAYERS = 2

DROPOUT = 0.1

# ============================================================
# DATASET CLASS
# ============================================================

class GPTDataset(Dataset):

    def __init__(self, tokens, block_size):

        self.tokens = tokens
        self.block_size = block_size

    def __len__(self):

        return len(self.tokens) - self.block_size

    def __getitem__(self, idx):

        x = torch.tensor(
            self.tokens[idx:idx+self.block_size],
            dtype=torch.long
        )

        y = torch.tensor(
            self.tokens[idx+1:idx+self.block_size+1],
            dtype=torch.long
        )

        return x, y

# ============================================================
# DATALOADERS
# ============================================================

train_dataset = GPTDataset(train_tokens, BLOCK_SIZE)

val_dataset = GPTDataset(val_tokens, BLOCK_SIZE)

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    pin_memory=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    pin_memory=True
)

# ============================================================
# MULTI HEAD ATTENTION
# ============================================================

class MultiHeadAttention(nn.Module):

    def __init__(self, embed_dim, num_heads, dropout=0.1):

        super().__init__()

        assert embed_dim % num_heads == 0

        self.embed_dim = embed_dim
        self.num_heads = num_heads

        self.head_dim = embed_dim // num_heads

        self.query = nn.Linear(embed_dim, embed_dim)
        self.key = nn.Linear(embed_dim, embed_dim)
        self.value = nn.Linear(embed_dim, embed_dim)

        self.fc_out = nn.Linear(embed_dim, embed_dim)

        self.dropout = nn.Dropout(dropout)

    def forward(self, x):

        B, T, C = x.shape

        Q = self.query(x)
        K = self.key(x)
        V = self.value(x)

        Q = Q.view(B, T, self.num_heads, self.head_dim).transpose(1,2)
        K = K.view(B, T, self.num_heads, self.head_dim).transpose(1,2)
        V = V.view(B, T, self.num_heads, self.head_dim).transpose(1,2)

        attention = (Q @ K.transpose(-2,-1)) / math.sqrt(self.head_dim)

        mask = torch.tril(torch.ones(T,T, device=x.device))

        attention = attention.masked_fill(mask == 0, float('-inf'))

        attention = torch.softmax(attention, dim=-1)

        attention = self.dropout(attention)

        out = attention @ V

        out = out.transpose(1,2).contiguous().view(B,T,C)

        out = self.fc_out(out)

        return out

# ============================================================
# FEED FORWARD NETWORK
# ============================================================

class FeedForward(nn.Module):

    def __init__(self, embed_dim, dropout=0.1):

        super().__init__()

        self.net = nn.Sequential(

            nn.Linear(embed_dim, 4 * embed_dim),

            nn.ReLU(),

            nn.Linear(4 * embed_dim, embed_dim),

            nn.Dropout(dropout)
        )

    def forward(self, x):

        return self.net(x)

# ============================================================
# TRANSFORMER BLOCK
# ============================================================

class TransformerBlock(nn.Module):

    def __init__(
        self,
        embed_dim,
        num_heads,
        dropout=0.1,
        use_norm=True,
        use_residual=True,
        use_ffn=True
    ):

        super().__init__()

        self.use_norm = use_norm
        self.use_residual = use_residual
        self.use_ffn = use_ffn

        self.attention = MultiHeadAttention(
            embed_dim,
            num_heads,
            dropout
        )

        self.ffn = FeedForward(embed_dim, dropout)

        self.norm1 = nn.LayerNorm(embed_dim)
        self.norm2 = nn.LayerNorm(embed_dim)

    def forward(self, x):

        attn = self.attention(x)

        if self.use_residual:
            x = x + attn
        else:
            x = attn

        if self.use_norm:
            x = self.norm1(x)

        if self.use_ffn:

            ffn_out = self.ffn(x)

            if self.use_residual:
                x = x + ffn_out
            else:
                x = ffn_out

            if self.use_norm:
                x = self.norm2(x)

        return x

# ============================================================
# GPT MODEL
# ============================================================

class GPTModel(nn.Module):

    def __init__(
        self,
        vocab_size,
        embed_dim,
        block_size,
        num_heads,
        num_layers,
        dropout=0.1,
        use_norm=True,
        use_residual=True,
        use_ffn=True
    ):

        super().__init__()

        self.token_embedding = nn.Embedding(
            vocab_size,
            embed_dim
        )

        self.position_embedding = nn.Embedding(
            block_size,
            embed_dim
        )

        self.blocks = nn.Sequential(*[

            TransformerBlock(
                embed_dim,
                num_heads,
                dropout,
                use_norm,
                use_residual,
                use_ffn
            )

            for _ in range(num_layers)

        ])

        self.ln_f = nn.LayerNorm(embed_dim)

        self.head = nn.Linear(embed_dim, vocab_size)

        self.dropout = nn.Dropout(dropout)

    def forward(self, x, targets=None):

        B, T = x.shape

        token_emb = self.token_embedding(x)

        positions = torch.arange(0, T, device=x.device)

        pos_emb = self.position_embedding(positions)

        out = token_emb + pos_emb

        out = self.dropout(out)

        out = self.blocks(out)

        out = self.ln_f(out)

        logits = self.head(out)

        loss = None

        if targets is not None:

            B, T, C = logits.shape

            logits = logits.view(B*T, C)

            targets = targets.view(B*T)

            loss = nn.functional.cross_entropy(
                logits,
                targets
            )

        return logits, loss

# ============================================================
# TRAIN FUNCTION
# ============================================================

def train_model(
    epochs=3,
    learning_rate=1e-3,
    num_layers=2,
    num_heads=2,
    use_norm=True,
    use_residual=True,
    use_ffn=True
):

    torch.cuda.empty_cache()
    gc.collect()

    model = GPTModel(
        vocab_size=VOCAB_SIZE,
        embed_dim=EMBED_DIM,
        block_size=BLOCK_SIZE,
        num_heads=num_heads,
        num_layers=num_layers,
        dropout=DROPOUT,
        use_norm=use_norm,
        use_residual=use_residual,
        use_ffn=use_ffn
    ).to(DEVICE)

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=learning_rate
    )

    scaler = torch.cuda.amp.GradScaler()

    train_losses = []
    val_losses = []

    for epoch in range(epochs):

        model.train()

        total_train_loss = 0

        loop = tqdm(train_loader)

        for x, y in loop:

            x = x.to(DEVICE)
            y = y.to(DEVICE)

            optimizer.zero_grad()

            with torch.cuda.amp.autocast():

                _, loss = model(x, y)

            scaler.scale(loss).backward()

            scaler.step(optimizer)

            scaler.update()

            total_train_loss += loss.item()

            loop.set_description(f"Epoch {epoch+1}")

            loop.set_postfix(loss=loss.item())

        avg_train_loss = total_train_loss / len(train_loader)

        model.eval()

        total_val_loss = 0

        with torch.no_grad():

            for x, y in val_loader:

                x = x.to(DEVICE)
                y = y.to(DEVICE)

                with torch.cuda.amp.autocast():

                    _, loss = model(x, y)

                total_val_loss += loss.item()

        avg_val_loss = total_val_loss / len(val_loader)

        train_losses.append(avg_train_loss)
        val_losses.append(avg_val_loss)

        print(
            f"\nEpoch {epoch+1}/{epochs}"
            f"\nTrain Loss: {avg_train_loss:.4f}"
            f"\nValidation Loss: {avg_val_loss:.4f}\n"
        )

    return model, train_losses, val_losses
```

```python
# ============================================================
# EXPERIMENT 3
# VARYING ATTENTION HEADS
# ============================================================

head_configs = [1, 2, 4]

head_results = {}

for heads in head_configs:

    print(f"\n================================================")
    print(f"TRAINING MODEL WITH {heads} ATTENTION HEADS")
    print(f"================================================\n")

    torch.cuda.empty_cache()
    gc.collect()

    model, train_loss, val_loss = train_model(
        epochs=3,
        learning_rate=1e-3,
        num_layers=2,
        num_heads=heads
    )

    head_results[heads] = {

        "train_loss": train_loss,
        "val_loss": val_loss
    }

    # --------------------------------------------------------
    # PLOT INDIVIDUAL RESULTS
    # --------------------------------------------------------

    plt.figure(figsize=(8,5))

    plt.plot(
        train_loss,
        marker='o',
        label='Train Loss'
    )

    plt.plot(
        val_loss,
        marker='o',
        label='Validation Loss'
    )

    plt.xlabel("Epoch")

    plt.ylabel("Loss")

    plt.title(f"{heads} Attention Heads")

    plt.legend()

    plt.grid(True)

    plt.show()

# ============================================================
# FINAL COMPARISON PLOT
# ============================================================

plt.figure(figsize=(10,6))

for heads in head_configs:

    plt.plot(
        head_results[heads]["val_loss"],
        marker='o',
        label=f'{heads} Heads'
    )

plt.xlabel("Epoch")

plt.ylabel("Validation Loss")

plt.title("Comparison of Attention Heads")

plt.legend()

plt.grid(True)

plt.show()

# ============================================================
# PRINT FINAL RESULTS
# ============================================================

print("\nFINAL RESULTS\n")

for heads in head_configs:

    final_train = head_results[heads]["train_loss"][-1]

    final_val = head_results[heads]["val_loss"][-1]

    print(f"{heads} Attention Heads")
    print(f"Final Train Loss: {final_train:.4f}")
    print(f"Final Validation Loss: {final_val:.4f}")
    print("-" * 40)

# ============================================================
# EXPERIMENT 3 ANALYSIS
# ============================================================

print("\nINFERENCE:\n")

print("1 Head  -> Learns limited relationships and context")

print("2 Heads -> Better contextual understanding and balanced learning")

print("4 Heads -> Richer feature learning but may increase overfitting")
```

```python
# ============================================================
# EXPERIMENT 4
# ABLATION STUDIES
# ============================================================

# Using only 2 epochs to reduce runtime

# ============================================================
# A. REMOVE LAYER NORMALIZATION
# ============================================================

print("\n================================================")
print("TRAINING WITHOUT LAYER NORMALIZATION")
print("================================================\n")

torch.cuda.empty_cache()
gc.collect()

norm_model, norm_train, norm_val = train_model(
    epochs=2,
    learning_rate=1e-3,
    num_layers=2,
    num_heads=2,
    use_norm=False
)

# ------------------------------------------------------------
# PLOT RESULTS
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

plt.plot(
    norm_train,
    marker='o',
    label='Train Loss'
)

plt.plot(
    norm_val,
    marker='o',
    label='Validation Loss'
)

plt.xlabel("Epoch")

plt.ylabel("Loss")

plt.title("Without Layer Normalization")

plt.legend()

plt.grid(True)

plt.show()

# ============================================================
# B. REMOVE RESIDUAL CONNECTIONS
# ============================================================

print("\n================================================")
print("TRAINING WITHOUT RESIDUAL CONNECTIONS")
print("================================================\n")

torch.cuda.empty_cache()
gc.collect()

res_model, res_train, res_val = train_model(
    epochs=2,
    learning_rate=1e-3,
    num_layers=2,
    num_heads=2,
    use_residual=False
)

# ------------------------------------------------------------
# PLOT RESULTS
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

plt.plot(
    res_train,
    marker='o',
    label='Train Loss'
)

plt.plot(
    res_val,
    marker='o',
    label='Validation Loss'
)

plt.xlabel("Epoch")

plt.ylabel("Loss")

plt.title("Without Residual Connections")

plt.legend()

plt.grid(True)

plt.show()

# ============================================================
# C. REMOVE FEED FORWARD NETWORK
# ============================================================

print("\n================================================")
print("TRAINING WITHOUT FEED FORWARD NETWORK")
print("================================================\n")

torch.cuda.empty_cache()
gc.collect()

ffn_model, ffn_train, ffn_val = train_model(
    epochs=2,
    learning_rate=1e-3,
    num_layers=2,
    num_heads=2,
    use_ffn=False
)

# ------------------------------------------------------------
# PLOT RESULTS
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

plt.plot(
    ffn_train,
    marker='o',
    label='Train Loss'
)

plt.plot(
    ffn_val,
    marker='o',
    label='Validation Loss'
)

plt.xlabel("Epoch")

plt.ylabel("Loss")

plt.title("Without Feed Forward Network")

plt.legend()

plt.grid(True)

plt.show()

# ============================================================
# FINAL COMPARISON PLOT
# ============================================================

plt.figure(figsize=(10,6))

plt.plot(
    norm_val,
    marker='o',
    label='No LayerNorm'
)

plt.plot(
    res_val,
    marker='o',
    label='No Residual'
)

plt.plot(
    ffn_val,
    marker='o',
    label='No FFN'
)

plt.xlabel("Epoch")

plt.ylabel("Validation Loss")

plt.title("Ablation Study Comparison")

plt.legend()

plt.grid(True)

plt.show()

# ============================================================
# FINAL RESULTS
# ============================================================

print("\nFINAL RESULTS\n")

print("WITHOUT LAYER NORMALIZATION")
print(f"Final Train Loss: {norm_train[-1]:.4f}")
print(f"Final Validation Loss: {norm_val[-1]:.4f}")
print("-" * 40)

print("WITHOUT RESIDUAL CONNECTIONS")
print(f"Final Train Loss: {res_train[-1]:.4f}")
print(f"Final Validation Loss: {res_val[-1]:.4f}")
print("-" * 40)

print("WITHOUT FEED FORWARD NETWORK")
print(f"Final Train Loss: {ffn_train[-1]:.4f}")
print(f"Final Validation Loss: {ffn_val[-1]:.4f}")
print("-" * 40)

# ============================================================
# ANALYSIS
# ============================================================

print("\nINFERENCE:\n")

print("Without LayerNorm -> Training becomes unstable")

print("Without Residual Connections -> Poor gradient flow and degraded learning")

print("Without FFN -> Reduced nonlinear learning capability")
```
